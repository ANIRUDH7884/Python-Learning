```python
# clean columns
df.columns = df.columns.str.strip().str.lower()

# encode categorical
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['car'] = le.fit_transform(df['car'])
df['model'] = le.fit_transform(df['model'])

# features & target
x = df[['car', 'volume', 'model', 'weight']]
y = df['co2']

# split & train
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
car_input = int(input("Enter the car name: "))
volume_input = float(input("Enter the volume: "))
model_input = int(input("Enter the model name: "))
weight_input = float(input("Enter the weight: "))

input_data = [[car_input, volume_input, model_input, weight_input]]

predicted_co2 = model.predict(input_data)

print(f"Predicted CO2 emission: {predicted_co2[0]}")

```
