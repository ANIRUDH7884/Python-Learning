```python
import matplotlib.pyplot as plt;

x = [10, 20, 30, 40]
y = [20, 30, 40, 60]

plt.title("Graph")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.plot(x , y, marker="p" , color="r")
plt.show()
```

```python
#vertical bar graph
import matplotlib.pyplot as plt

x = ["A", "B", "C", "D"]
Y = [10, 20, 30, 40]

plt.bar(x,y , color="r", width=0.8)
plt.show()
```

```python
#Horizontal bar graph
import matplotlib.pyplot as plt

x = ["A", "B", "C", "D"]
Y = [10, 20, 30, 40]

plt.barh(x,y , color="r")
plt.show()
```

```python
#Histogram

data = [20000, 25000, 30000, 35000, 40000, 45000, 50000, 55000, 60000, 65000, 75000]

#ploting the histogram
plt.hist(data , bins=5, edgecolor = "black")

plt.title("Car price Range")
plt.xlabel = "price"
plt.ylable = "frequency"

plt.show()
```

```python
import matplotlib.pyplot as plt

# data
labels = ["Petrol", "Diesel", "CNG", "Electric"]
sizes = [50, 20, 10, 20]
colors = ["red", "yellow", "green", "blue"]

# plotting
plt.pie(
    sizes,                     # ✅ correct
    labels=labels,
    colors=colors,
    explode=(0.1, 0, 0, 0),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Resource Consumption")  # ✅ correct
plt.show()
```
