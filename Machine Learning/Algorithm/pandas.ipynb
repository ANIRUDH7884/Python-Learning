```python
#Pandas working with the datasets

import pandas as pd

data = {
    "Name": ["Anu", "Rahul", "Arun", "Meera", "John"],
    "Age": [22, 25, 23, 24, 26],
    "Salary": [15000, 25000, 20000, 30000, 18000],
    "Department": ["IT", "HR", "IT", "Sales", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

```python

print(df['Name'])

print(df[df["Age"] > 23])

print(df[df["Salary"] > 20000])

print(df[df["Department"] == 'IT'])

```

```python
difference = df[(df["Department"] == "IT")& (df['Salary'] > 20000)]
print(difference)
```

```python
or_checking = df[(df['Department'] == "Sales") | (df['Salary'] > 20000)]
print(or_checking)
```

```python
df.sort_values(by="Salary", ascending=False)
```

```python
df.sort_values(by='Age', ascending= True)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python

import pandas as pd

data = {
    "Name": ["Anu", "Rahul", "Arun", "Meera", "John"],
    "Age": [22, 25, None, 24, 26],
    "Salary": [15000, None, 20000, 30000, 18000],
    "Department": ["IT", "HR", "IT", None, "HR"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df_cleaned = df.dropna()
print(df_cleaned)
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean)
print(df)
```

```python
df['Age'] = df['Age'].fillna('Unknown')
print(df)
```

```python
df['Department'] = df['Department'].fillna('Unknown')
print(df)
```
