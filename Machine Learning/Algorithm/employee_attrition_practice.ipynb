```python
#Imports

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sbn

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

```python
#Load Data
dt = pd.read_csv('/content/employee_attrition_practice.csv')
```

```python
#Preprocess The Object Values into Integer
le = LabelEncoder()
dt['overtime'] = le.fit_transform(dt['overtime'])
dt['department'] = le.fit_transform(dt['department'])
dt['salary_level'] = le.fit_transform(dt['salary_level'])
```

```python
#MinMax scaler
```

```python
dt
```

```python
#Assign the values

x = dt.drop('left', axis = 1)
y = dt['left']

print(x)
print(y)
```

```python
#Train Test Split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
print(x_train)
print(x_test)
print(y_train)
print(y_test)
```

```python
#MinMax scaler

scaler = MinMaxScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#train Model

model = SVC(
    kernel='linear',
    class_weight='balanced'
)

model.fit(x_train, y_train)
```

```python
#ModeL Predict
y_prediction = model.predict(x_test)
print(y_prediction)
```

```python
#Comparison

comparison = pd.DataFrame({

    'Actual Value' : y_test,
    'Predicted Value' : y_prediction
})

print(comparison)
```

```python
#Accuracy Score
Accuracy = accuracy_score(y_test, y_prediction)
print(Accuracy)
```

```python
#Confusion matrix
cm = confusion_matrix(y_test, y_prediction)
print(cm)
```

```python
#classification Report
cr = classification_report(y_test, y_prediction)
print(cr)
```
