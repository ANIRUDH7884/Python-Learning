```python
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
Data = pd.read_csv('/content/DATA (1).csv')
Data.head()
```

```python
Data.tail()
```

```python
Data = Data.drop("Unnamed: 5", axis=1)
```

```python
#Encode using Label Encoder
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

Data['Car'] = le.fit_transform(Data['Car'])
Data['Model'] = le.fit_transform(Data['Model'])

Data.head()
```

```python
#Assign Values

x = Data[['Car', 'Volume', 'Weight']]
y = Data['CO2']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(y_pred)
```

```python
comparison = pd.DataFrame({
    'Actual CO2 Emission' : y_test.values,
    'Predicted CO2 Emission' : y_pred
})

comparison['differnce'] = (comparison['Actual CO2 Emission'] - comparison['Predicted CO2 Emission'])

print(comparison.round(2))
```

```python
#Evaluation
MAE = mean_absolute_error(y_test,y_pred)
MSE = mean_squared_error(y_test,y_pred)
R2 = r2_score(y_test,y_pred)

print('Mean Absolute Error : ' , MAE)
print('Mean Squared Error : ', MSE)
print('R2 Score : ', R2)
```
