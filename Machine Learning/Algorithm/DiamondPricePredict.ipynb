```python
#Imports
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

#Import Data

Diamond_Data = pd.read_csv('/content/diamond.csv')
```

```python
Diamond_Data.head()
```

```python
Diamond_Data.info()
```

```python
Diamond_Data.isnull().sum()
```

```python
#Encode Datas
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

Diamond_Data['Cut'] = le.fit_transform(Diamond_Data['Cut'])
Diamond_Data['Color'] = le.fit_transform(Diamond_Data['Color'])
Diamond_Data['Clarity'] = le.fit_transform(Diamond_Data['Clarity'])
Diamond_Data['Polish'] = le.fit_transform(Diamond_Data['Polish'])
Diamond_Data['Symmetry'] = le.fit_transform(Diamond_Data['Symmetry'])
Diamond_Data['Report'] = le.fit_transform(Diamond_Data['Report'])
```

```python
Diamond_Data
```

```python
#Assign Values

x = Diamond_Data[['Carat Weight',	'Cut', 'Color',	'Clarity',	'Polish',	'Symmetry',	'Report']]
y = Diamond_Data['Price']
```

```python
#Spliting

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#model Training

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
#predict
out_predict = model.predict(x_test)
print(out_predict.round(2))
```

```python
#Comparison
comparison = pd.DataFrame({
    'Actual Price' : y_test.values,
    'Prediction Price' : out_predict
})

comparison['Error'] = (comparison['Actual Price'] - comparison['Prediction Price'])
print(comparison.round(2))
```

```python
from google.colab import files

comparison.to_csv("Diamond_Price_Comparison.csv", index=False)

files.download("Diamond_Price_Comparison.csv")
```

```python
MAE = mean_absolute_error(y_test,out_predict)
MSE = mean_squared_error(y_test,out_predict)
R2 = r2_score(y_test,out_predict)

print('Mean Absolute Error : ' , MAE)
print('Mean Squared Error : ', MSE)
print('R2 Score : ', R2)
```

```python
#Live Prediction
carat_weight = float(input("Enter Carat Weight: "))
cut = int(input("Enter Cut: "))
color = int(input("Enter Color: "))
clarity = int(input("Enter Clarity: "))
polish = int(input("Enter Polish: "))
symmetry = int(input("Enter Symmetry: "))
report = int(input("Enter Report: "))

user_input = [[
    carat_weight, cut, color, clarity,
    polish, symmetry, report
]]

prediction = model.predict(user_input)

print("Predicted Diamond Price:", round(prediction[0], 2))
```
