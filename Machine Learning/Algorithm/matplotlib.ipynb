```python
import matplotlib.pyplot as plt

# data
labels = ["Petrol", "Diesel", "CNG", "Electric"]
sizes = [50, 20, 10, 20]
colors = ["red", "yellow", "green", "blue"]

# plotting
plt.pie(
    sizes,                     # ✅ correct
    labels=labels,
    colors=colors,
    explode=(0.1, 0, 0, 0),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Resource Consumption")  # ✅ correct
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('/content/student_score_prediction.csv')

df = pd.DataFrame(data)
df
```

```python

plt.plot(
    df['study_hours'],
    df['exam_score'],
    marker='p',
    color='b',
    label='Exam Score'
)

plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.legend()
plt.show()
```

```python
plt.scatter(
    df['study_hours'],
    df['exam_score'],
    marker='p',
    color='b',
    label='Exam Score'
)

plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.legend()
plt.show()
```

```python
plt.bar(
    df['study_hours'],
    df['exam_score'],
    color='g',
    label='Exam Score'
)

plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.legend()
plt.show()
```

```python

plt.hist(df["exam_score"], bins=5, color="orange", edgecolor="black")

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")

plt.show()
```

```python

import matplotlib.pyplot as plt

month = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [10000 , 15000, 12000, 20000, 25000, 30000]

plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.plot(month, sales, color='r', marker='p')
plt.title("Sales Over Months")
plt.show()
```

```python
import matplotlib.pyplot as plt

month = ['Electronics', 'Cosmetics', 'Fashion',
         'Wearables', 'Bags', 'Mobiles']

sales = [600000, 15000, 120000, 200000, 25000, 300000]

plt.bar(month, sales, color='b', width=0.5, label='Sales')

plt.xlabel('Product Category')
plt.ylabel('Sales (₹)')
plt.title('Sales by Product Category')
plt.legend()
plt.show()
```
